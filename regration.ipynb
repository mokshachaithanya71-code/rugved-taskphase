```python
from collections import deque
import pandas as pd
a=pd.read_csv("c:\\ml\\co2.csv")
x=a["Engine Size(L)"]
y=a["CO2 Emissions(g/km)"]
x=list(x)
y=list(y)
n=len(y)
grabo=deque(maxlen=n)
grab1=deque(maxlen=n)
#er=deque(maxlen=n)
b=True
j=0
bo=3
b1=2
p=0.01
r=deque(maxlen=60)
while b:
    sum1=0
    g11=0
    goo=0
    for i in range(0,n):
        t=bo+b1*x[i]
        e=t-y[i]
        ##er.append(e)
        go=e*1
        grabo.append(go)
        g1=e*x[i]
        grab1.append(g1)
        sum1=sum1+e*e
    g11=sum(grab1)/n
    goo=sum(grabo)/n
    b1=b1-p*g11
    bo=bo-p*goo
    r.appendleft(sum1)
    if len(r)==1:
        continue
    err=r[j+1]-r[j]
    if err>0 and err<1:
        break
print("the weights are ",b1," and bias is ",bo)
    
```
