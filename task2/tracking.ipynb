```python
import cv2 as cv
import numpy as np
from collections import deque
d=cv.VideoCapture("C:\\photoes\\Ball_Tracking.mp4")
pts=deque(maxlen=60)
while True:
    istrue,f=d.read()
    if not istrue:
        print("no video")
        break
    c=f.copy()
    h=cv.cvtColor(f,cv.COLOR_BGR2HSV)
    l=np.array([29,86,6])
    u=np.array([64,255,255])
    m=cv.inRange(h,l,u)
    o=m.copy()
    m=cv.erode(m,None,2)
    m=cv.dilate(m,None,2)
    cont,heir=cv.findContours(m,cv.RETR_EXTERNAL,cv.CHAIN_APPROX_SIMPLE)
    if len(cont)==0:
        continue
    a=max(cont,key=cv.contourArea)
    t=cv.moments(a)
    if t["m00"]!=0:
        cent=(int(t["m10"]/t["m00"]),int(t["m01"]/t["m00"]))
    pts.appendleft(cent)
    for i in range(0,len(pts)-1):
        if pts[i]is None or pts[i+1]is None:
            continue
        cv.line(c,pts[i],pts[i+1],(0,0,255),4)
    cv.imshow("t",c)
    if cv.waitKey(20)&0xFF==ord("b"):
        break
d.release()##stops reading frames
cv.destroyAllWindows()## to close all opencv functions used
```
